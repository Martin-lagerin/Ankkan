```python
class Wares:

    def __init__(self , name: str , price: int , brand: str , expiry: int , stock: int) ->None:

        self.name = name
        self.price = price
        self.brand = brand
        self.expiry = expiry
        self.stock = stock

    def __add__ (self ,other: 'wares' ) -> None:

        print ('haha')
        print ( self.brand + other.brand)
        print(self.name + other.name)

        return f'in total we have {self.stock + other.stock} {self.name} and {other.name}'


    def __mul__ (self, other:'wares') -> str:

        '''den här class metoden kommer kosta totalt'''

        total_self_cost = self.stock * self.price
        total_other_cost = other.stock * other.price

        print(f' {self.name} kostar totalt {total_self_cost}')
        print(f' {other.name} kostar totalt {total_other_cost}')

        return f' total cost {total_self_cost + total_other_cost}'
```
