```python
def longest_word(apa):
    """
    This function takes a list of words and returns the longest word in the list.
    """
    longest = ""                        # testade 0 och ändra till while och olika for loopar.
    
    for mapa in apa:                    # ändra variablar och funktionsnamn
        if len(longest) > len(mapa):    # shifta < teknet och skriva in longest först sen word som kopplar till det angivna namnet i forloopen ,<? testa mig fram
            longest = mapa              # ändra 'mapa' till 'word' så alla har samma funktionsnamn, kan ändra longest till vad som helst. men ska inte tas bort 
    
    return longest
```
