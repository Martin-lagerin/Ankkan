```python
class Flygplan:
 
    def __init__ (self, brand: str, model: str, year: int) -> None:

        self.brand = brand
        self.model = model
        self.year = year

    def presentation(self) ->None:

        return f'this {self.model} is from {self.year} and the company is {self.brand}'

    def trash (self) -> None:

        life_lenght = self.year + 50

        return (f' this {self.model} expeirs in {life_lenght}')
```

```python
info = Flygplan('ML-wings', 'M-1992', 1992 )
```

```python
print(info.presentation())
print(info.trash())
```

```python
class Militärflygplan(Flygplan):

    def __init__ (self,brand: str, model: str, year: int, air_height:float)->None:

        super().__init__(brand,model,year)
        self.__air_height= air_height

    @property
    def air_height(self)->float:
        return self.__air_height

    @air_height.setter
    def air_height(self, air_height: float | int) ->None:
        if isinstance (air_height, float):
            self.__air_height = air_height
        else:
            raise TypeError (f'must be an int or float')

    def hight (self)->str:

        return f'when {self.model} fly, the air drop is {self.air_height}cm high'
```

```python
young_info = Militärflygplan('ML-wings', 'M-2015', 2015, 50)
```

```python
print(young_info.presentation())
print(young_info.hight())
```

```python
class Passagerarflygplan(Flygplan):

    def __init__ ( self,brand: str, model: str, year: int, seats: int)->None:

        super().__init__(brand,model,year)
        self.seats= seats

    def seats_availible(self, my_seats: int) ->None:

        av_seats = 500

        return f' in this {self.model} its 500 setas, now is {my_seats} occupide, its {av_seats - my_seats} seats left'
```
