```python
class Varor:

    

    def __init__ (self, namn: str, pris: int, märke: str, datum: float, saldo: int):

        ''' en funktion som kollar vad som finns i en affär
        #attribut
        namn str : namnet på varan
        pris int: priset på varan
        märke str: märket på varan
        giltighetsdatum float : utgångsdatum på varan
        lagersaldo int: hur många varor av produkten 
        '''
        self.namn = namn
        self.pris = pris
        self.märke = märke
        self.datum= datum

        self.lagersaldo = saldo
    #@property
    #def self.__lagersaldo(self):
        #return


    def vilken_vara_går_ut(self):

        print(f'den här {self.namn} går ut {self.datum}')

    def hur_många(self):

        print(f'av den här varan: {self.varan} har vi {self.saldo} st')

    def discount (self,rabatt:int):

        if not 0 <= rabatt <= 100:
            raise ValueError ('rabatt måste vara mellan 0 till 100')

        rabbat_belopp = self.pris * (rabatt/100)
        beräknat_pris= (self.pris - rabbat_belopp)
        return (f' varan: {self.namn} har rabbaterats {rabbat_belopp}kr. alltså {beräknat_pris}')
 

    def allt(self):

        print(f'varan är: {self.namn} - varan kostar: {self.pris}kr - varan kommer ifrån: {self.märke} - varan går ut den: {self.datum} - finns :{self.lagersaldo}st av varan') # första variablen self.lagersaldo 

    def __repr__(self):

        ''' returnerar en vara på ett uinambigos/uppenbart sätt'''

        return f"Varor(namn='{self.namn}', pris={self.pris}, märke='{self.märke}', datum='{self.datum}', saldo='{self.lagersaldo}'"


        #return (f" Varor(namn={self.namn}, pris= {self.pris}, märke= {self.märke} , datum={self.datum}, saldo={self.lagersaldo}" )
```

```python
class Circle:

    def __init__(self, radius: int ,color:str) -> None:

        self.__radius = radius
        self.__color = color

    @property
    def radius(self) -> int:
        return self.__radius

    @radius.setter
    def radius(self, radius:int):
       
        if isinstance(radius,int):
            self.__radius = radius
        
        else:
            raise TypeError ('måste vara en int')

```

```python
class Rektangel:

    def __init__ (self, längd:int, höjd:int)->None:

        self.__längd = längd
        self.__höjd = höjd

    @property 
    def höjd(self) -> int:

        return self.__höjd

    @property
    def längd(self) -> int:

        return self.__längd

    @längd.setter
    def längd(self,value:int):

        if isinstance(value, int):

            if value > 0:
                self.__längd = value

            else:
                raise ValueError('måste vara poitiv längd int')

        else:
            raise TypeError ( 'måste vara en längd int inte str')

    @höjd.setter
    def höjd(self,value:int):

        if isinstance(value, int):

            if value > 0:
                self.__höjd = value

            else:
                raise ValueError('måste vara poitiv höjd int')

        else:
            raise TypeError ( 'måste vara en höjd int inte str')
```
