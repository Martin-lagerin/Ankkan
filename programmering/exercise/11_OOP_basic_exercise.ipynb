````python
class Converter:

    def __init__(self, inch_to_cm=2.54,foot_to_meters= 0.3048 , pound_to_kg=0.453592):

        self.inch_to_cm = inch_to_cm
        self.foot_to_meters = foot_to_meters
        self.__pound_to_kg = pound_to_kg


    @property
    def pound_to_kg(self):
        return self.__pound_to_kg

    @pound_to_kg.setter
    def pound_to_kg(self,value):
        if isinstance(value):
            if value < 0:
                self.__pound_to_kg = value
            else:
                raise ValueError ('måste vara ett int eller float, pound property')
        else:
            raise TypeError ('måste vara en int eller float, pound property 2')

    

    def inch_in_cm(self, inch):

        if not isinstance(inch,(int,float)):
            raise TypeError ('värdet måste vara int eller float, inch')
        if inch < 0:
            raise ValueError ('måste vara ett poitivt tal, inch')
       
        cm = inch * self.inch_to_cm

        print(f' {inch} inch är {cm:.2f} i cm')

    
    def foot_in_meters(self, foot:int):

        if not isinstance(foot,(int,float)):
            raise TypeError (' måste vara ett en int eller float, foot') 
        if foot < 0:
            raise ValueError ('måste vara ett postivt tal , foot')

        meter = foot * self.foot_to_meters

        print(f' {foot} foot är {meter:.2f} i meter')

    def pound_in_kg(self,pound):

        if not isinstance(pound,(int,float)):
            raise TypeError ('måste vara ett int eller float , pound')
        if pound < 0:
            raise ValueError ('måste vara ett postivt tal , pound')

        kg = pound * self.pound_to_kg

        print(f' {pound} pound är {kg:.2f} i kg')

    def __repr__(self):

        return f'en inch till cm= {self.inch_to_cm}cm , en foot till meter ={self.foot_to_meters}m ,en pound till kg= {self.pound_to_kg}kg'
````

````python
Converter= Converter()
Converter.inch_in_cm(20)
Converter.foot_in_meters(20)
Converter.pound_in_kg(20)
````

````python
Converter.pound_to_kg = -80
````

````python
class Person:
    
    def __init__(self,namn : str , ålder: int, email :str) -> None:

        if not isinstance(namn ,str):
            raise TypeError ('namn måste vara en sträng')
        if not isinstance(ålder, (int,float)):
            raise ValueError ('skriv en int eller float')

        self.namn = namn
        self.age = ålder
        self.email = email



    def __repr__ (self):

        return f' namn : {self.namn} , ålder :{self.age} , email: {self.email}'

    def hello (self, namn , ålder, email):

        return f' hej jag heter {self.namn}, jag är {self.age}år gamal och min email är {self.email}'
````
