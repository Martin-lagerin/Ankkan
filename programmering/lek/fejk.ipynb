```python
class Varor:
    """
    En klass som representerar en vara i en affär.
    """

    def __init__(self, namn: str, pris: int, märke: str, datum: str, lagersaldo: int):
        """
        Initierar en ny vara.

        Args:
            namn (str): Namnet på varan.
            pris (int): Priset på varan.
            märke (str): Märket på varan.
            datum (str): Giltighetsdatum på varan.
            lagersaldo (int): Hur många varor av produkten finns i lager.
        """
        self.namn = namn
        self.pris = pris
        self.märke = märke
        self.datum = datum
        self.saldo = lagersaldo


    def vilken_vara_går_ut(self):
        """Visar när varan går ut."""
        print(f'Den här varan ({self.namn}) går ut {self.datum}')

    def hur_många(self):
        """Visar lagersaldot."""
        print(f'Av {self.namn} har vi {self.saldo} st i lager.')

    def allt(self):
        """Visar all information om varan."""
        print(f'{self.namn} - {self.pris} - {self.märke} - {self.datum} - {self.saldo}')

    def __repr__(self):
        """Otvetydig representation av instansen."""
        return f"Varor(namn='{self.namn}', pris={self.pris}, märke='{self.märke}', datum='{self.datum}', saldo={self.saldo})"
```
