```python
x = np.arange(1,11)
p = np.array([0.05, 0.10, 0.15, 0.05, 0.05, 0.05, 0.05, 0.05, 0.40, 0.05])
```

```python
# A) Sannolikhetsfördelning

plt.bar (x, p, color="#750582")
plt.xlabel ('x')
plt.ylabel ('P(X = x)', rotation=0, labelpad=25)
plt.title ('Sannolikhetsfördelningen')
plt.show()
```

```python
# B) Kumulativa fordelningssfunktionen F(x)

# pi = den aktuella sannolikheten från listan p
# summa = den kumulativa summan hittills
# F = listan med alla kumulativa sannolikheter

F = []
summa = 0

for pi in p:
    summa += pi
    F.append(summa)

plt.step(x, F, where='post', color='#890909')
plt.xlabel('x')
plt.ylabel('F(x)', rotation=0, labelpad=15)
plt.title('Kumulativ fördelningsfunktion')
plt.ylim(0, 1.05)
plt.show()
```

```python
# C) Berakna E(X), V (X) samt S(X).​

# E(X) = sum(x * P(X = x))
# V(X) = sum(x^2 * P(X = x)) - [E(X)]^2
# S(X) = sqrt(V(X))

E_X = np.sum(x * p)
V_X = np.sum(x**2 * p) - E_X**2
S_X = np.sqrt(V_X)

print(f'E(X) = {E_X:.2f}')
print(f'V(X) = {V_X:.2f}')
print(f'S(X) = {S_X:.2f}')
```
