Here's the boilerplate (starter code) for a class called Faction that you'll be working on. The instances of this class will represent the different competing factions.

```python
import math
import matplotlib.pyplot as plt
import random
import time


class Faction:

    # A class representing factions fruit war

    def __init__ ( self , name: str , apples: int, oranges: int , bananas: int , strawberries: int) -> None:
    #--------------------------
        #Atributes
        self.name = name
        self.apples = apples
        self.oranges = oranges
        self.bananas = bananas
        self.strawberries = strawberries

    #----------------------------
        # Properties och Setter
    #----------------------------
    # Name
    @property
    def name(self) -> str:
        return self._name
    

    @name.setter
    def name(self,name: str) -> None:
        ''' Setter for validatet name'''
        try:
            if not isinstance(name, str):
                raise TypeError ('Faction name must be a string.')
            self._name = name
        except TypeError as e:
            print(f'Invalid name: {e}')
            self._name = 'unknown'
   
    #--------
    # Apples
    @property
    def apples(self) -> int:
        return self._apples
    
    @apples.setter
    def apples(self, value: int) -> None:
        """Setter for validated integer """
        try:
            if not isinstance(value, int):
                raise TypeError('Apples must be an integer.')
            if not 0 <= value <= 200:
                raise ValueError('Apples must be between 0 and 200.')
            self._apples = value
        except (TypeError, ValueError) as e:
            print(f'Apples invalid {e}')
            self._apples = 0

    #--------
    # Oranges
    @property
    def oranges(self) -> int:
        return self._oranges
    
    @oranges.setter
    def oranges(self, value: int)->None:
        """Setter for validated integer """
        try:
            if not isinstance(value, int):
                raise TypeError ('Oranges must be an integer')
            if not 0 <= value <=200:
                raise ValueError ('Oranges must be between 0 and 200')
            self._oranges = value
        except (TypeError,ValueError) as e:
            print(f'Oranges is invalid {e}')
            self._oranges = 0

    #--------
    # Bananas
    @property
    def bananas(self) -> int:
        return self._bananas
    
    @bananas.setter
    def bananas(self, value: int)-> None:
        """Setter for validated integer """
        try:
            if not isinstance(value, int):
                raise TypeError ('Bananas must be an integer')
            if not 0 <= value <=200:
                raise ValueError ('Bananas must be between 0 and 200')
            self._bananas = value
        except (TypeError,ValueError) as e:
            print(f'Bananas is invalid {e}')
            self._bananas = 0

    #--------
    # Strawberries
    @property
    def strawberries(self) -> int:
        return self._strawberries
    
    @strawberries.setter
    def strawberries(self, value: int)-> None:
        """Setter for validated integer """
        try:
            if not isinstance(value, int):
                raise TypeError ('strawberries must be an integer')
            if not 0<= value <= 200:
                raise ValueError ('strawberries must be between 0 and 200')
            self._strawberries = value
        except (TypeError,ValueError) as e:
            print(f'Strawberries is invalid {e}')
            self._strawberries = 0

    #--------------------
        #Metodhs
    #-------------------- 
    def inventory(self) -> int:
        ''' Return the total amount of the fruits'''
        return self.apples + self.oranges + self.bananas + self.strawberries
    
    #--------------
    def status(self) -> int:
        ''' Shows the satus of were each faction stands. '''

        total = self.inventory()
        if total <= 150:
            return f'This aint nothin but a scratch! {self.name} might be struggling for the moment, but will be back with vengance!'
        elif total <= 500:
            return f'The proud {self.name} is VERY eager to toss some apples! Where are the opponents?!'
        else:
            return f'{self.name} is GODLIKE! Their position is super dominant!'

    #--------------
    def visualize_resources(self) -> None:

        ''' Shows a bar chart of how much fruits each faction has. '''

        fruits = ['apples', 'oranges', 'bananas','strawberries' ]
        fruit_amount = [self.apples, self.oranges, self.bananas, self.strawberries]

        colors = []

        for q in fruit_amount:
            if q >= 100:
                colors.append('green')
            elif q > 50:
                colors.append('orange')
            else:
                colors.append('red')

        plt.figure(figsize=(4,4))
        plt.bar(fruits,fruit_amount, color = colors)
        plt.title(f'{self.name} Fruit amount')
        plt.xlabel('Fruit type')
        plt.ylabel('Amount')
        plt.ylim(0,220)
        plt.show()

    #--------------
    def organical_growth(self) -> None:

        '''Simulate organic growth for each type of fruit'''

        fruits = [('apples', self.apples), ('oranges', self.oranges), ('bananas', self.bananas), ('strawberries', self.strawberries)]

        for name, amount in fruits:
            if amount < 200:
                growth = random.randint(1,5)
                new_amount = min (amount + growth, 200)
                if name == 'apples':
                    self.apples = new_amount
                elif name == 'oranges':
                    self.oranges = new_amount
                elif name == 'bananas':
                    self.bananas = new_amount 
                elif name == 'strawberries':
                    self.strawberries = new_amount
                print(f"Glorious {self.name} has harvested {new_amount - amount} {name}, to a new total of {new_amount}!")
            else: 
                print(f'Glorious {self.name} is already at maximum capacity for {name}!')
                    
    #--------------
    def __mul__(self, other) -> None:

        ''' Operator to simulate an attack between factions '''

        print(f'{self.name} is attacking {other.name}')
        time.sleep(1.5)

        # Chek if other faction has recources
        if other.inventory() == 0:
            print(f'{other.name} has no resources left. choose another faction')

        # Determine if attack succeeds (70%)
        if random.randint(1,100) > 70:
            print(f'{other.name} defended good! the attack failed!')
            return None
        
        # Attack succees
        print('The attack was succssefull')
        time.sleep(1.5)

        # chek for critical hit
        critical = random.randint(1,100) <= 10
        loot_percent = 0.4 if critical else 0.2

        if critical:
            print(f'critical strike! {self.name} will loot {int(loot_percent * 100)}% of all resources')
        else:
            print(f'{self.name} will loot {int(loot_percent * 100)}% of all resources')
        time.sleep(1.5)

        # stolen fruits
        fruits = ['apples', 'oranges', 'bananas', 'strawberries']
        for fruit in fruits:
            defender_amount = getattr(other, fruit)
            attack_amount = getattr(self, fruit)

            # humilitating blow
            if 0 <= defender_amount <= 5:
                loot = defender_amount
                print(f'{self.name} stole all {defender_amount} {fruit} from {other.name}.')
            else:
                loot= int(defender_amount * loot_percent)

            # Defender loses loot
            new_defender_amount = max(defender_amount - loot, 0)
            setattr(other, fruit, new_defender_amount)
            
            # Attacker gains loot
            space_left = 200 - attack_amount
            gained = min(loot, space_left) 
            setattr(self, fruit, attack_amount + gained)

            if gained < loot:
                print(f'{self.name} looted {gained} {fruit}, but {loot - gained} were lost in the cahos of war.' )
            else:
                print(f'{self.name} looted {gained} {fruit} from {other.name}.')

            time.sleep(1.2)

        print(f'{self.name} WON the Battle against {other.name}! Sorry, maybe next time.')
        return None


    #--------------
    def trade_offer(self) -> None:
        
        pass

    #--------------
    def __repr__(self) -> str:

        '''Returns amonts of all the fruits'''

        return f' Name = {self.name}, Apples = {self.apples}pc, Oranges = {self.oranges}pc, Bananas = {self.bananas}pc and Strawberries = {self.strawberries}pc.'

```
