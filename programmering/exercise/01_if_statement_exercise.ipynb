## 1. Check sign (*)

Ask the user to input a number and check if this number is **positive**, **negative** or **zero** and print it out.

```python


while True:

    try:
        number = int(input('skriv en siffra, vi ska hitta en jämn siffra'))
    
        if number == 0:
            print ('siffran är noll')
        

        elif number % 2==0:
            print ('siffran är positiv')
            break

        else:
            print ('siffran är ojämn') 
    except ValueError:
        print ('skriv en giltig siffra')
    
        
```
