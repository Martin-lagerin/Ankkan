## 2. Dice simulation (*)

Simulate 1000000 dice rolls and save the number of ones, twos, ..., sixes in a dictionary. Then print them out in the terminal. (*)

<details>

<summary>Answer</summary>

```
1: 166481
2: 166844
3: 166154
4: 167400
5: 167260
6: 165861
```

</details>


````python
import random 

dice_roll = [random.randint(1,6) for _ in range(1000000)]

result= {}

for i in range(1,7):

    result[i] = dice_roll.count(i)

print(result)

for key,value in result.items():

    to_print= f'\n{key}\n{value}'

print(to_print)


    
````

````python
import random

rolls = 1000000

count_dice = {i: 0 for i in range(1,7)}

for x in range(rolls):
    roll=random.randint(1,6)
    count_dice[roll] += 1

for side in range(1,7):
    count = count_dice[side]
    percent = count / rolls
    print(f"{side}: {count} ({percent:.2f}%)")



````
