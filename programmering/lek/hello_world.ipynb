```python
# Slumpar ett tal mellan 1 och 100
slumptal = random.randint(1, 100)

print("Jag tänker på ett tal mellan 1 och 100... kan du gissa vilket?")

# Starta en loop som körs tills man gissar rätt
gissning = None
antal_försök = 0

while gissning != slumptal:
    gissning = int(input("Din gissning: "))
    antal_försök += 1

    if gissning < slumptal:
        print("För lågt! 🔽")
    elif gissning > slumptal:
        print("För högt! 🔼")
    else:
        print(f"Rätt gissat! 🎉 Talet var {slumptal}. Du behövde {antal_försök} försök.")
```

```python
import random

# Skapa en lista med 10 tärningskast
rolls = [random.randint(1, 6) for x in range(10)]
print("Original rolls:", rolls)

# a) Sortera i stigande ordning
asc = sorted(rolls)
print("Ascending order:", asc)

# b) Sortera i fallande ordning
desc = sorted(rolls, reverse=True)
print("Descending order:", desc)

# c) Hitta max och min
print("Maximum:", max(rolls))
print("Minimum:", min(rolls))
```
