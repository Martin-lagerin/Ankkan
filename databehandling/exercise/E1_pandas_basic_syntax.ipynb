````python
import pandas as pd

cities = { 
        'kommun': ['Malmö', 'Stockholm', 'Uppsala', 'Göteborg'],
        'population': [347949, 975551, 233839, 583056]
}
````

````python
# A
städer = pd.DataFrame(cities)
städer
````

````python
# B
mask = [stad == 'Göteborg' for stad in städer['kommun']]
städer[mask]
````

````python
# C
gradvis = städer.sort_values(by='population', ascending=False)
gradvis
````

````python
# D
sorterad = gradvis.iloc[:3]
sorterad
````

````python
# E     10379295
total_population = 10379295
städer['population %']= (gradvis['population']/total_population).round(3) * 100
nya = städer.sort_values(by='population', ascending=False)
nyas = nya.reset_index(drop=True)
nyas
````

````python
df = pd.read_excel('../listor/komtopp50_2020.xlsx', sheet_name='Totalt',skiprows=6)
df
````

````python
df.columns = [ 'Rang 2020', 'Rang 2019', 'Kommun', 'Folkmängd 2020', 'Folkmängd 2019', 'Förändring']
df.dropna()
````

````python
df['Rang 2019'] = df['Rang 2019'].astype(int)
df['Rang 2020'] = df['Rang 2020'].astype(int)
df['Folkmängd 2020'] = df['Folkmängd 2020'].astype(int)
df['Folkmängd 2019'] = df['Folkmängd 2019'].astype(int)
df['Förändring'] = ((df['Folkmängd 2020'] - df['Folkmängd 2019']) / df['Folkmängd 2019']) * 100
````

````python

df_sorted = df.sort_values('Folkmängd 2020', ascending=False)
df_sorted.head()
````

````python
df_smallest = df_sorted.tail(5)
df_smallest
````

````python
total2019=df['Folkmängd 2019'].sum()
total2020 = df['Folkmängd 2020'].sum()

print(f'totalen 2019 : {total2019} \n'
      f'total 2020 : {total2020}')
print(f'förändring :', total2020 - total2019)
````

````python
import matplotlib.pyplot as plt
largest = df.head(5)

plt.figure(figsize=(10,5))
plt.bar (largest['Kommun'],largest['Folkmängd 2020'])
plt.bar (df_smallest['Kommun'], df_smallest['Folkmängd 2020'])
plt.legend()
plt.show()
````
